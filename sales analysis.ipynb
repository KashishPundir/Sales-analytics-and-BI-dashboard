Q5: Does 20% of customers generate 80% revenue? (Pareto Analysis)

````python
# Calculate Revenue by Client_ID

client_revenue = (
    df.groupby('Client_ID')['Total_Bill']
      .sum()
      .sort_values(ascending=False)
)

client_revenue.head()
````

````python
# Calculate Revenue Share

client_revenue_df = client_revenue.reset_index()

client_revenue_df.columns = [
    'Client_ID',
    'Revenue'
]

client_revenue_df['Revenue_%'] = (
    client_revenue_df['Revenue']
    /
    client_revenue_df['Revenue'].sum()
) * 100
````

````python
#  Calculate Cumulative Revenue Percentage

client_revenue_df['Cumulative_Revenue_%'] = (
    client_revenue_df['Revenue_%']
    .cumsum()
)

client_revenue_df.head()
````

````python
# Revenue generated by each customer
customer_revenue = (
    df.groupby('Client_ID')['Revenue']
      .sum()
      .sort_values(ascending=False)
)

# Number of customers
n_customers = len(customer_revenue)

# Top 20% count
top_20_count = int(n_customers * 0.20)

# Revenue generated by top 20%
top_20_revenue = customer_revenue.iloc[:top_20_count].sum()

# Total revenue
total_revenue = customer_revenue.sum()

# Percentage of total revenue
top_20_percentage = (top_20_revenue / total_revenue) * 100

print("Total customers:", n_customers)
print("Top 20% customers:", top_20_count)
print("Top 20% revenue:", top_20_revenue)
print("Revenue contribution:", top_20_percentage, "%")
````

It follows the Pareto principle(80/20 rule), where the top 20% of clients generate approximately 88% of total revenue.

# Geographic  Analysis

````python
# Revenue by City

city_revenue = (
    df.groupby(
        'Client City'
    )
    ['Total_Bill']

    .sum().sort_values(ascending=False).head()
)
print(city_revenue)
````

````python
# Customer count

city_customers = (
    df.groupby(
        'Client City'
    )
    ['Client_ID']
    .nunique().sort_values(ascending=False).head()
)

city_customers
````

````python
# Revenue by city
city_revenue = (
    df.groupby('Client City')['Total_Bill']
      .sum()
      .sort_values(ascending=False)
      .reset_index()
)

# Revenue percentage
city_revenue['Revenue_%'] = (
    city_revenue['Total_Bill']
    / city_revenue['Total_Bill'].sum()
) * 100

# Cumulative percentage
city_revenue['Cumulative_%'] = city_revenue['Revenue_%'].cumsum()

# Cities contributing first 50% revenue
top_50_cities = city_revenue[
    city_revenue['Cumulative_%'] <= 50
]

# Include the city that crosses the 50% threshold
if top_50_cities['Cumulative_%'].max() < 50:
    top_50_cities = city_revenue[
        city_revenue.index <= len(top_50_cities)
    ]

print(top_50_cities)
````

````python
# Calculate Revenue Per Client by City

city_client_analysis = (
    df.groupby('Client City')
      .agg(
          Total_Revenue=('Total_Bill', 'sum'),
          Unique_Clients=('Client_ID', 'nunique')
      )
)

city_client_analysis['Revenue_Per_Client'] = (
    city_client_analysis['Total_Revenue']
    / city_client_analysis['Unique_Clients']
)

city_client_analysis = city_client_analysis.sort_values(
    'Revenue_Per_Client',
    ascending=False
)
````

````python
city_client_analysis = city_client_analysis[
    city_client_analysis['Unique_Clients'] >= 10
]

city_client_analysis.sort_values(
    'Revenue_Per_Client',
    ascending=False
).head(10)
````

The analysis shows that Hosur has the highest revenue generated per client, indicating the presence of high-value customers in the city. Faridabad stands out as a key market because it combines a large customer base with high revenue per client, making it one of the most valuable and stable markets for the company. In contrast, cities such as Bangalore and Chennai have lower average revenue per client, suggesting opportunities to increase customer spending through targeted business strategies and service expansion.
